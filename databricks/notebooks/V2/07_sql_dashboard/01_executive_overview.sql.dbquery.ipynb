```sql
-- ============================================================
-- Executive Manufacturing Dashboard
-- ============================================================

SELECT

    SUM(operations_completed) AS total_operations,

    ROUND(AVG(pass_rate), 2) AS average_pass_rate,

    ROUND(AVG(average_cycle_time_sec), 2) AS average_cycle_time_sec,

    ROUND(AVG(average_force_kn), 2) AS average_force_kn,

    CURRENT_TIMESTAMP() AS generated_timestamp

FROM smip_v2.manufacturing.machine_kpis;

--- Production Summary
SELECT

    SUM(work_orders_created) AS work_orders,

    SUM(executions_started) AS executions,

    SUM(products_started) AS products_started

FROM smip_v2.manufacturing.production_kpis;

--- Quality Summary
SELECT

    SUM(tests_completed) AS tests_completed,

    SUM(tests_passed) AS tests_passed,

    SUM(tests_failed) AS tests_failed,

    ROUND(AVG(pass_rate),2) AS quality_pass_rate

FROM smip_v2.manufacturing.quality_kpis;


--- Material Summary

SELECT

    SUM(materials_scanned) AS materials_scanned,

    SUM(successful_scans) AS successful_scans,

    SUM(failed_scans) AS failed_scans,

    ROUND(AVG(scan_success_rate),2) AS scan_success_rate

FROM smip_v2.manufacturing.material_kpis;


--- Packaging Summary

SELECT

    SUM(packages_completed) AS packages_completed,

    ROUND(AVG(shipment_readiness_rate),2) AS shipment_readiness_rate

FROM smip_v2.manufacturing.packaging_kpis;
```
