```python
"""
SMIP V2

Bronze Streaming Table

Reads Bronze manufacturing events from a Unity Catalog Volume.

Author:
Sumanth Vempalle

Version:
2.2.0
"""

from pyspark import pipelines as dp

from pyspark.sql.functions import (
    col,
)


@dp.table(
    name="manufacturing_events",
    comment="Bronze manufacturing events.",
    table_properties={
        "quality": "bronze",
        "pipelines.autoOptimize.managed": "true",
    },
)

@dp.expect_or_drop(
    "valid_payload",
    "raw_event IS NOT NULL",
)

def manufacturing_events():

    return (

        dp.read_stream(
            "bronze_raw_events"
        )

        .select(

            col("value").cast("string").alias(
                "raw_event"
            )

        )

    )
```
