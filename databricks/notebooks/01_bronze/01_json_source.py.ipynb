```python
"""
SMIP V2

Bronze JSON Source

Reads raw manufacturing events from a Unity Catalog Volume.

Author:
Sumanth Vempalle

Version:
2.1.0
"""

from pyspark import pipelines as dp


RAW_DATA_PATH = (
    "/Volumes/smip_v2/manufacturing/raw_data/"
)


@dp.table(
    name="bronze_raw_events",
    comment="Raw manufacturing events from Unity Catalog Volume.",
    table_properties={
        "quality": "bronze",
        "pipelines.autoOptimize.managed": "true",
    },
)
def bronze_raw_events():

    return (

        spark.readStream

        .format("text")

        .load(RAW_DATA_PATH)

    )
```
