```python
"""
02_fact_quality.py

Manufacturing Quality Fact Table

Business Grain:
    One completed quality inspection/test.

Sources:
    quality_events
    product_dimension

Target:
    fact_quality

Author:
Sumanth Vempalle

Version:
2.3.0
"""

import dlt

from pyspark.sql.functions import col


# ============================================================
# Quality Fact Table
# ============================================================

@dlt.table(
    name="fact_quality",
    comment="Manufacturing Quality Fact Table.",
    table_properties={
        "quality": "gold",
        "pipelines.autoOptimize.managed": "true",
    },
)
def fact_quality():

    quality = dlt.read(
        "quality_events"
    )

    products = dlt.read(
        "product_dimension"
    )

    return (

        quality.alias("q")

        .join(

            products.alias("pd"),

            on="product_code",

            how="left",

        )

        .select(

            # ====================================================
            # Event
            # ====================================================

            col("q.event_id"),

            col("q.event_timestamp"),

            col("q.event_version"),

            # ====================================================
            # Business Keys
            # ====================================================

            col("q.plant_code"),

            col("q.execution_id"),

            col("q.serial_number"),

            col("q.product_code"),

            # ====================================================
            # Product Dimension
            # ====================================================

            col("pd.product_name"),

            col("pd.family"),

            col("pd.rated_voltage_kv"),

            col("pd.routing_version"),

            # ====================================================
            # Quality Measures
            # ====================================================

            col("q.test_result_id"),

            col("q.test_program_id"),

            col("q.test_name"),

            col("q.target_value"),

            col("q.measured_value"),

            col("q.unit"),

            col("q.result"),

            # ====================================================
            # Audit
            # ====================================================

            col("q.silver_processing_timestamp"),

        )

    )
```
