```python
"""
04_fact_packaging.py

Manufacturing Packaging Fact Table

Business Grain:
    One completed packaging event.

Sources:
    packaging_events
    product_dimension

Target:
    fact_packaging

Author:
Sumanth Vempalle

Version:
2.3.0
"""

import dlt

from pyspark.sql.functions import col


# ============================================================
# Packaging Fact Table
# ============================================================

@dlt.table(
    name="fact_packaging",
    comment="Manufacturing Packaging Fact Table.",
    table_properties={
        "quality": "gold",
        "pipelines.autoOptimize.managed": "true",
    },
)
def fact_packaging():

    packaging = dlt.read(
        "packaging_events"
    )

    products = dlt.read(
        "product_dimension"
    )

    return (

        packaging.alias("p")

        .join(

            products.alias("pd"),

            on="product_code",

            how="left",

        )

        .select(

            # ====================================================
            # Event
            # ====================================================

            col("p.event_id"),

            col("p.event_timestamp"),

            col("p.event_version"),

            # ====================================================
            # Manufacturing Keys
            # ====================================================

            col("p.plant_code"),

            col("p.execution_id"),

            col("p.serial_number"),

            col("p.product_code"),

            # ====================================================
            # Product Dimension
            # ====================================================

            col("pd.product_name"),

            col("pd.family"),

            col("pd.rated_voltage_kv"),

            col("pd.routing_version"),

            # ====================================================
            # Packaging
            # ====================================================

            col("p.package_id"),

            col("p.package_type"),

            col("p.package_weight_kg"),

            col("p.package_length_mm"),

            col("p.package_width_mm"),

            col("p.package_height_mm"),

            col("p.packaging_status"),

            # ====================================================
            # Audit
            # ====================================================

            col("p.silver_processing_timestamp"),

        )

    )
```
