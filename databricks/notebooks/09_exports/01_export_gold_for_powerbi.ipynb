### Imports

```python
from pathlib import Path

from pyspark.sql import DataFrame
```

### Configuration

```python
# ============================================================================
# Export Location
# ============================================================================

EXPORT_PATH = "/Volumes/smip/exports/powerbi"
```

### Load Gold Tables

```python
production = spark.table("smip.gold.production_summary")

quality = spark.table("smip.gold.quality_summary")

oee = spark.table("smip.gold.oee_summary")

traceability = spark.table("smip.gold.traceability_summary")

executive = spark.table("smip.gold.executive_summary")
```

### Export Function

```python
from pyspark.sql import DataFrame

def export_csv(df: DataFrame, name: str):

    output_path = f"{EXPORT_PATH}/{name}"

    (
        df.coalesce(1)
          .write
          .mode("overwrite")
          .option("header", True)
          .csv(output_path)
    )

    print(f"✓ Exported {name}")
```

### Export

```python
print("=" * 80)
print("Exporting Gold Tables")
print("=" * 80)

export_csv(production, "production_summary")

export_csv(quality, "quality_summary")

export_csv(oee, "oee_summary")

export_csv(traceability, "traceability_summary")

export_csv(executive, "executive_summary")
```
