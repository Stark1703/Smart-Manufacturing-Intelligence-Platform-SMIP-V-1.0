### Imports

```python
import os
import shutil
```

### Configuration

```python
# ============================================================================
# Paths
# ============================================================================

EXPORT_ROOT = "/Volumes/smip/exports/powerbi"

FINAL_EXPORT = "/Volumes/smip/exports/powerbi_exports"

os.makedirs(FINAL_EXPORT, exist_ok=True)
```

### Helper Function

```python
def prepare_csv(folder_name):

    source_folder = os.path.join(EXPORT_ROOT, folder_name)

    files = os.listdir(source_folder)

    csv_file = next(
        f for f in files
        if f.startswith("part-") and f.endswith(".csv")
    )

    source = os.path.join(source_folder, csv_file)

    destination = os.path.join(
        FINAL_EXPORT,
        f"{folder_name}.csv"
    )

    shutil.copy2(source, destination)

    print(f"✓ {folder_name}.csv")
```

### Execute

```python
print("=" * 80)
print("Preparing Power BI Files")
print("=" * 80)

prepare_csv("production_summary")

prepare_csv("quality_summary")

prepare_csv("oee_summary")

prepare_csv("traceability_summary")

prepare_csv("executive_summary")
```

### Validation

```python
print()

print("Generated Files")

print("-" * 80)

for file in sorted(os.listdir(FINAL_EXPORT)):

    print(file)
```

### Summary
